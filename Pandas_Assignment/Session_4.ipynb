TASK:1 1.
Create a DataFrame named 'movies' with columns: 'title', 'rating', and 'release_year' for at least 6 popular Bollywood films. Use sort_values() to display the movies sorted by 'rating' from highest to lowest.

```python
import pandas as pd
```

```python
data = {
    'title': [
        '3 Idiots',
        'Dangal',
        'PK',
        'Bajrangi Bhaijaan',
        'Lagaan',
        'Taare Zameen Par'
    ],
    'rating': [8.4, 8.3, 8.1, 8.1, 8.1, 8.3],
    'release_year': [2009, 2016, 2014, 2015, 2001, 2007]
}
```

```python
movies = pd.DataFrame(data)
```

```python
sorted_movies = movies.sort_values(
    by='rating',
    ascending=False
)
```

TASK :3.
Reset the index of a DataFrame containing a list of trending YouTube videos (columns: 'video_title', 'views', 'likes'), and make sure the old index is not added as a column in the result.<br><br><em><strong>Constraint:</strong> Use the drop=True argument in reset_index().</em>

```python
data = {
    'video_title': [
        'Python Tutorial',
        'Cricket Highlights',
        'Funny Cat Videos',
        'Bollywood Songs',
        'Travel Vlog'
    ],
    'views': [1500000, 2500000, 800000, 1200000, 950000],
    'likes': [85000, 120000, 45000, 70000, 50000]
}
```

```python

youtube_videos = pd.DataFrame(data, index=[10, 20, 30, 40, 50])
```

```python
print("Original DataFrame:")
print(youtube_videos)
```

```python
result = youtube_videos.reset_index(drop=True)
```

```python
print("\nDataFrame after resetting index:")
print(result)
```

TASK:4.
You have a DataFrame of Zomato restaurant names and their ratings. Use reindex() to rearrange the rows so that your favorite restaurant appears first, followed by the rest in any order.

```python
data = {
    'restaurant': [
        'Dominos',
        'McDonalds',
        'Haldirams',
        'Pizza Hut',
        'Burger King'
    ],
    'rating': [4.2, 4.0, 4.4, 4.1, 3.9]
}
```

```python
zomato = pd.DataFrame(data)
zomato = zomato.set_index('restaurant')
```

```python
print("Original DataFrame:")
print(zomato)
```

```python
result = zomato.reindex([
    'Haldirams',
    'Dominos',
    'Pizza Hut',
    'McDonalds',
    'Burger King'
])
```

```python
print("Reindexed DataFrame:")
print(result)
```

TASK:5.
Take a DataFrame of Flipkart products with columns 'product_name', 'price', and 'discount'. First, sort the products by 'discount' descending, then reset the index so it starts from 0 and is sequential after sorting.

```python
data = {
    'product_name': ['Laptop', 'Mobile', 'Headphones', 'Smartwatch', 'Tablet'],
    'price': [60000, 20000, 3000, 5000, 25000],
    'discount': [10, 25, 40, 15, 30]
}
```

```python
df = pd.DataFrame(data)
```

```python
df = df.sort_values(by='discount', ascending=False)
```

```python
df = df.reset_index(drop=True)
```

```python
print(df)
```
