TASK-1.
Download a dataset of IPL cricket matches (CSV) and use pandas to group the data by 'team' to find the total runs scored by each team. Display the result as a DataFrame.

```python

import pandas as pd
```

```python
df = pd.read_csv("ipl_team_runs.csv")
```

```python
total_runs = df.groupby("team", as_index=False)["runs"].sum()
```

```python
print(total_runs)
```

TASK-2.
Given a DataFrame of Zomato food orders with columns ['city', 'restaurant', 'order_amount'], use groupby to calculate the average order_amount for each restaurant in each city using the agg() function.

```python
df = pd.DataFrame({
    'city': [
        'Ahmedabad', 'Ahmedabad', 'Mumbai',
        'Mumbai', 'Ahmedabad', 'Mumbai',
        'Delhi', 'Delhi'
    ],
    'restaurant': [
        'Dominos', 'Dominos', 'Dominos',
        'Pizza Hut', 'Pizza Hut', 'Pizza Hut',
        'McDonalds', 'McDonalds'
    ],
    'order_amount': [300, 500, 400, 600, 700, 800, 250, 450]
})
```

```python
result = df.groupby(['city', 'restaurant']).agg(
    average_order_amount=('order_amount', 'mean')
).reset_index()
```

```python
print(result)
```

TASK-3.
Take a Flipkart-style product sales DataFrame with columns ['category', 'region', 'sales', 'units_sold'], and apply multiple aggregations: for each (category, region) pair, calculate the total sales and the average units_sold using groupby and agg().

```python
df = pd.DataFrame({
    'category': [
        'Electronics', 'Electronics', 'Clothing',
        'Clothing', 'Electronics', 'Clothing',
        'Electronics', 'Clothing'
    ],
    'region': [
        'West', 'West', 'North', 'North',
        'South', 'West', 'South', 'North'
    ],
    'sales': [50000, 30000, 15000, 20000, 40000, 10000, 25000, 18000],
    'units_sold': [10, 6, 30, 40, 8, 20, 5, 35]
})
```

```python
result = df.groupby(['category', 'region']).agg(
    total_sales=('sales', 'sum'),
    average_units_sold=('units_sold', 'mean')
).reset_index()
```

```python
print(result)
```

TASK-4.
Refactor the code that groups a DataFrame by 'region' and sums 'revenue' so it also returns the maximum and minimum revenue per region in the same result.<br><br><em><strong>Hint:</strong> Use agg() with a dictionary to specify multiple aggregation functions for the 'revenue' column.</em>

```python
df = pd.DataFrame({
    'region': ['West', 'North', 'West', 'South', 'North', 'South', 'West'],
    'revenue': [50000, 30000, 70000, 40000, 45000, 60000, 25000]
})
```

```python
result = df.groupby('region').agg({
    'revenue': ['sum', 'max', 'min']
})
```

```python
result.columns = ['total_revenue', 'max_revenue', 'min_revenue']
```

```python
print(result)
```

TASK-5.
Suppose you have a DataFrame of Spotify song streams with columns ['artist', 'genre', 'streams']. Use groupby and agg() to find both the total and average streams for each genre, then sort the result by total streams in descending order.<br><br><em><strong>Constraint:</strong> Your final DataFrame should have genres as the index and columns for total and average streams.</em>

```python
df = pd.DataFrame({
    'artist': ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'genre': ['Pop', 'Rock', 'Pop', 'Hip-Hop', 'Rock', 'Hip-Hop', 'Pop'],
    'streams': [1000000, 800000, 1500000, 600000, 1200000, 900000, 500000]
})
```

```python
result = df.groupby('genre').agg({
    'streams': ['sum', 'mean']
})
```

```python
result.columns = ['total_streams', 'average_streams']
```

```python
result = result.sort_values(by='total_streams', ascending=False)
```

```python
print(result)
```
