TASK-1.
Use pd.concat to combine two DataFrames representing January and February orders from a Swiggy-style food delivery app, stacking them row-wise to create a single DataFrame with all orders.

```python
import pandas as pd
```

```python
january_orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'customer_name': ['Amit', 'Priya', 'Rahul'],
    'restaurant': ['Pizza Hut', 'Dominos', 'Burger King'],
    'order_amount': [500, 350, 250]
})
```

```python
february_orders = pd.DataFrame({
    'order_id': [104, 105, 106],
    'customer_name': ['Neha', 'Karan', 'Riya'],
    'restaurant': ['McDonalds', 'Pizza Hut', 'Dominos'],
    'order_amount': [300, 600, 400]
})
```

```python
all_orders = pd.concat(
    [january_orders, february_orders],
    axis=0,
    ignore_index=True
)
```

```python
print(all_orders)
```

TASK-3.
Create a DataFrame showing daily song streams for three users on Spotify for one week, then use the pivot function to reshape the data so each row is a user and each column is a day.

```python
df = pd.DataFrame({
    'user': [
        'Amit', 'Amit', 'Amit', 'Amit', 'Amit', 'Amit', 'Amit',
        'Priya', 'Priya', 'Priya', 'Priya', 'Priya', 'Priya', 'Priya',
        'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul', 'Rahul'
    ],
    'day': [
        'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'
    ] * 3,
    'streams': [
        20, 25, 30, 22, 35, 40, 45,
        15, 18, 25, 20, 30, 38, 42,
        10, 12, 18, 15, 22, 28, 32
    ]
})
```

```python
print("Original DataFrame:")
print(df)
```

```python
pivot_df = df.pivot(
    index='user',
    columns='day',
    values='streams'
)
```

```python
pivot_df = pivot_df[
    ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
     'Friday', 'Saturday', 'Sunday']
]
```

```python
print("\nPivoted DataFrame:")
print(pivot_df)
```

TASK-4.
Given a DataFrame of Zomato restaurant orders (with columns: restaurant, month, total_amount), use pivot_table to calculate the average order amount per restaurant for each month.

```python
df = pd.DataFrame({
    'restaurant': [
        'Dominos', 'Dominos', 'Dominos',
        'McDonalds', 'McDonalds', 'McDonalds',
        'Pizza Hut', 'Pizza Hut', 'Pizza Hut'
    ],
    'month': [
        'January', 'January', 'February',
        'January', 'February', 'February',
        'January', 'February', 'February'
    ],
    'total_amount': [300, 500, 400, 250, 350, 450, 600, 500, 700]
})
```

```python
pivot_df = pd.pivot_table(
    df,
    index='restaurant',
    columns='month',
    values='total_amount',
    aggfunc='mean'
)
```

```python
print(pivot_df)
```

TASK-5.
Download three monthly sales CSV files for a mock Myntra store (e.g., sales_jan.csv, sales_feb.csv, sales_mar.csv), then use pd.concat to combine them into one DataFrame and display the total sales per product.<br><br><em><strong>Hint:</strong> Use groupby on the product column after concatenation to sum sales.</em>

```python
jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")
```

```python
all_sales = pd.concat([jan, feb, mar], ignore_index=True)
```

```python
print("Combined Sales Data:")
print(all_sales)
```

```python
total_sales = all_sales.groupby("product", as_index=False)["sales"].sum()
```

```python
print("\nTotal Sales Per Product:")
print(total_sales)
```
