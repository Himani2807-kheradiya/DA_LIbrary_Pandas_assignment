TASK-3.
Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.

```python
data = {
    'name': ['Restaurant A', 'Restaurant B', 'Restaurant C',
             'Restaurant D', 'Restaurant E'],
    'rating': [4.5, 4.2, 3.8, 4.7, 4.0],
    'votes': [1200, 900, 400, 1500, 650],
    'avg_cost': [800, 600, 450, 1000, 500]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Zomato Restaurants DataFrame:")
print(df)
```

```python
correlation = df['rating'].corr(df['votes'])
```

```python
print("\nCorrelation coefficient between rating and votes:",
      round(correlation, 3))
```
