TASK-1.
Create a Pandas DataFrame with a column 'username' containing 5 Instagram usernames in mixed case (e.g., 'CoolDude', 'foodieQueen', etc.), then use str.lower to convert all usernames to lowercase and print the result.

```python
import pandas as pd
```

```python
data = {
    'username': ['CoolDude', 'foodieQueen', 'TravelLover', 'TECH_GURU', 'FashionStar']
}
```

```python
df = pd.DataFrame(data)
```

```python
df['username'] = df['username'].str.lower()
```

```python
print(df)
```

TASk-2.
Given a Pandas Series of song titles from your Spotify playlist, use str.replace to remove all occurrences of the word 'Remix' from the titles and display the cleaned list.

```python
songs = pd.Series([
    'Perfect Remix',
    'Shape of You Remix',
    'Blinding Lights',
    'Levitating Remix',
    'Stay Remix'
])
```

```python
cleaned_songs = songs.str.replace('Remix', '', regex=False).str.strip()
```

```python
print(cleaned_songs)
```

TASK -3.
You have a DataFrame with a 'review' column containing user reviews from Zomato. Use str.contains to filter and display only the reviews that mention the word 'delivery'.<br><br><em><strong>Hint:</strong> Remember to handle case sensitivity so that 'Delivery' and 'delivery' are both matched.</em>

```python
data = {
    'review': [
        'Fast delivery and tasty food',
        'The food was delicious',
        'Delivery was very late',
        'Excellent service and quick Delivery',
        'The restaurant was clean',
        'I am happy with the delivery service'
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
filtered_reviews = df[df['review'].str.contains('delivery', case=False, na=False)]
```

```python
print(filtered_reviews)
```

TASK _4.
Take a Pandas Series of Flipkart product categories in the format 'Electronics/Mobiles/Smartphones', 'Home/Kitchen/Appliances', etc. Use str.split to extract just the main category (the part before the first '/') and create a new column with these values.

```python
categories = pd.Series([
    'Electronics/Mobiles/Smartphones',
    'Home/Kitchen/Appliances',
    'Fashion/Men/Clothing',
    'Electronics/Computers/Laptops',
    'Home/Furniture/Sofas'
])
```

```python
df = pd.DataFrame({'Category': categories})
```

```python
df['MainCategory'] = df['Category'].str.split('/').str[0]
```

```python
print(df)
```

TASK _5.
Given a DataFrame of YouTube video titles, use str.upper to convert all titles to uppercase, then chain str.replace to substitute every space with an underscore.<br><br><em><strong>Constraint:</strong> Do this in a single line of code for the transformation.</em>

```python
data = {
    'Title': [
        'Python Tutorial for Beginners',
        'Learn Pandas in Python',
        'How to Make a Website',
        'Data Science Full Course',
        'Machine Learning Basics'
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
df['Title'] = df['Title'].str.upper().str.replace(' ', '_', regex=False)
```

```python
print(df)
```
