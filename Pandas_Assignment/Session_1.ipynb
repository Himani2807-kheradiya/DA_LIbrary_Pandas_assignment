 TASK -3:
Given a list of follower counts for 5 Instagram influencers: [1200, 54000, 780, 150000, 32000], create a pandas Series called 'followers', then print the influencer(s) with the highest and lowest follower counts.<br><br><em><strong>Hint:</strong> Use Series.idxmax() and Series.idxmin() to find the indices.</em>

```python
followers = pd.Series([1200, 54000, 780, 150000, 32000])

print("Followers:")
print(followers)

highest = followers.idxmax()
lowest = followers.idxmin()

print("Highest followers:", followers[highest])
print("Highest index:", highest)

print("Lowest followers:", followers[lowest])
print("Lowest index:", lowest)
```
