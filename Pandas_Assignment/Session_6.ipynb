TASK :2.
Given a pandas DataFrame of Flipkart product reviews with columns: product_id, user_id, rating, review_text, identify and print the number of duplicate reviews (where product_id and user_id are both the same) before and after using drop_duplicates().

```python
data = {
    'product_id': [101, 102, 101, 103, 102, 104],
    'user_id': [1, 2, 1, 3, 2, 4],
    'rating': [5, 4, 3, 5, 4, 2],
    'review_text': [
        'Excellent product',
        'Good quality',
        'Average product',
        'Very nice',
        'Good quality',
        'Not satisfied'
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python

print("Original DataFrame:")
print(df)
```

```python
duplicates_before = df.duplicated(
    subset=['product_id', 'user_id']
).sum()
```

```python
print("\nNumber of duplicate reviews before removing:", duplicates_before)
```

```python
df_cleaned = df.drop_duplicates(
    subset=['product_id', 'user_id'],
    keep='first'
)
```

```python
duplicates_after = df_cleaned.duplicated(
    subset=['product_id', 'user_id']
).sum()
```

```python
print(df_cleaned)
```

```python
print("\nNumber of duplicate reviews after removing:", duplicates_after)
```

TASK:3.
Use the IQR method to detect outliers in the 'order_amount' column of a Zomato orders DataFrame and print the order IDs that are considered outliers.<br><br><em><strong>Hint:</strong> Calculate Q1, Q3, and IQR, then filter orders outside the [Q1 - 1.5*IQR, Q3 + 1.5*IQR] range.</em>

```python
data = {
    'order_id': [101, 102, 103, 104, 105, 106, 107, 108],
    'order_amount': [250, 300, 350, 400, 450, 500, 550, 5000]
}
```

```python

df = pd.DataFrame(data)
```

```python
print("Zomato Orders DataFrame:")
print(df)
```

```python
Q1 = df['order_amount'].quantile(0.25)
```

```python
Q3 = df['order_amount'].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
lower_bound = Q1 - 1.5 * IQR
```

```python
upper_bound = Q3 + 1.5 * IQR
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df['order_amount'] < lower_bound) |
    (df['order_amount'] > upper_bound)
]
```

```python
print("\nOutlier Order IDs:")
print(outliers['order_id'].tolist())
```

TASK :4.
For a DataFrame containing daily IPL match ticket sales (columns: date, tickets_sold), calculate the Z-score for each day's tickets_sold, and print the dates where the absolute Z-score is greater than 2.<br><br><em><strong>Constraint:</strong> Use pandas and scipy.stats.zscore for calculations.</em>

```python
data = {
    'date': [
        '2026-04-01', '2026-04-02', '2026-04-03',
        '2026-04-04', '2026-04-05', '2026-04-06',
        '2026-04-07', '2026-04-08'
    ],
    'tickets_sold': [1000, 1200, 1100, 1050, 1150, 1080, 1120, 5000]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original DataFrame:")
print(df)
```

```python
mean = df['tickets_sold'].mean()
```

```python
std = df['tickets_sold'].std(ddof=0)
```

```python
df['z_score'] = (df['tickets_sold'] - mean) / std
```

```python
outliers = df[df['z_score'].abs() > 2]
```

```python
print("\nDates with absolute Z-score greater than 2:")
print(outliers[['date', 'tickets_sold', 'z_score']])
```

TASK :5.
Use ChatGPT or Copilot to generate a pandas function that finds and returns all outlier transaction amounts in a Paytm wallet DataFrame using the Z-score method, then test the function with sample data.

```python
def find_outliers(df):
    mean = df['transaction_amount'].mean()
    std = df['transaction_amount'].std(ddof=0)

   
    if std == 0:
        return df.iloc[0:0]

   
    z_scores = (df['transaction_amount'] - mean) / std


    return df[z_scores.abs() > 2]
```

```python
data = {
    'transaction_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'user_name': [
        'Amit', 'Priya', 'Rahul', 'Neha', 'Karan',
        'Riya', 'Vikas', 'Pooja', 'Arjun', 'Meera'
    ],
    'transaction_amount': [
        100, 150, 200, 120, 180,
        250, 130, 170, 160, 5000
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Paytm Wallet Transactions:")
print(df)
```

```python
outliers = find_outliers(df)
```

```python
print("\nOutlier Transactions:")
print(outliers)
```

```python
print("\nOutlier Transaction Amounts:")
print(outliers['transaction_amount'].tolist())
```
